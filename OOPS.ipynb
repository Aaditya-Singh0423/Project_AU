Constructor is a built-in function in python which is used to initialise a variable in class

```python
class Bank:
    def __init__(self,account_number,balance,name):                     #Constructor  --> __init__
        self.account_number=account_number                              #Attributes
        self.balance=balance                                            #self is used to access attributes and its the object itslef
        self.name=name
```

```python
c1=Bank(1,22000,"Aaditya")
c2=Bank(2,20000,"Chirag")
```

```python
c1.name
```

```python
c2.balance
```

```python
class Animal:
    def __init__(self,name,age,number_of_legs):
        self.name=name
        self.age=age
        self.number_of_legs=number_of_legs
    def run(self):
        print("I am running")
    def eat(self):
        print("I am eating")
    def sleep(self):
        print("I am sleeping")
    def play(self):
        print("I am playing")
    def walk(self):
        print("I am walking")
    def fly(self):
        print("I am flying")
```

```python
class Fraction:
    def __init__(self,num,den):                  #__init__ is used to initialize the object
        self.num=num
        self.den=den
    def __str__(self):                            #__str__ is used to print the object
        return f"{self.num}/{self.den}"
    def __add__(self,other):                      #__add__ is used to add two objects
        new_num=self.num*other.den+self.den*other.num
        new_den=self.den*other.den
        return Fraction(new_num,new_den)
    def __sub__(self,other):
        new_num=self.num*other.den-self.den*other.num
        new_den=self.den*other.den
        return Fraction(new_num,new_den)
    def __mul__(self,other):
        new_num=self.num*other.num
        new_den=self.den*other.den
        return Fraction(new_num,new_den)
    def __truediv__(self,other):
        new_num=self.num*other.den
        new_den=self.den*other.num
        return Fraction(new_num,new_den)
```

```python
f1=Fraction(2,3)
```

```python
f2=Fraction(1,2)
```

```python
print(f1,f2)
```

```python
print(f1+f2)
```

```python
print(f1-f2)
```
